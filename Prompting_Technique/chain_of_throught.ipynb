```python
from dotenv import load_dotenv
import os
from openai import OpenAI


load_dotenv()

client  = OpenAI()

model = 'gpt-4o-mini'

def ask(prompt,system=None,temperature = 0):
    """Send One prompt and return the reply text. temperature=0 keep answere stable."""
    message = []
    if system:
        message.append({"role":"system","content":system})
    message.append({"role":"user","content":prompt})
    response = client.chat.completions.create(
            model=model,messages=message,temperature=temperature
        )

    return response.choices[0].message.content.strip()



def chain_of_thought():
    print("\n" + "=" * 60)
    print("2. CHAIN OF THOUGHT")
    print("=" * 60)
 
    question = (
        "A bakery had 23 cookies. 20 were sold in the morning. "
        "Then the baker made 6 more. 4 more were sold in the afternoon. "
        "Each cookie costs $3. How much money did the bakery make today, "
        "and how many cookies are left?"
    )

    direct = ask(question + "\nGive only the final answer, no explanation.")
    print("(a) WITHOUT CoT:\n", direct)
 
    cot = ask(
        question
        + "\nThink step by step. Write each step, then give the answer on a last line starting with 'Final:'."
    )
    print("\n(b) WITH CoT:\n", cot)
 
    print("\nCorrect answer: sold 20 + 4 = 24, earned 24 x $3 = $72; left 23 - 20 + 6 - 4 = 5 cookies.")
    print("Notice: with CoT every step is visible, so mistakes are easy to spot.")

    
chain_of_thought()
```
