```python
from dotenv import load_dotenv
import os
from openai import OpenAI
from collections import Counter


load_dotenv()

client  = OpenAI()

model = 'gpt-4o-mini'

def ask(prompt,system=None,temperature = 0):
    """Send One prompt and return the reply text. temperature=0 keep answere stable."""
    message = []
    if system:
        message.append({"role":"system","content":system})
    message.append({"role":"user","content":prompt})
    response = client.chat.completions.create(
            model=model,messages=message,temperature=temperature
        )

    return response.choices[0].message.content.strip()



question = (
    "A bakery had 23 cookies. 20 were sold in the morning. The baker made 6 more. "
    "4 more were sold in the afternoon. Each cookie costs $3. "
    "How much money did the bakery make? Think step by step, "
    "then give only the number on the last line as 'Final: <number>'."
)

answers = []
for i in range(5):
    reply = ask(question, temperature=0.7)        # temperature > 0 so each run reasons differently
    final_line = reply.strip().splitlines()[-1]     # e.g. "Final: 72"
    answer = final_line.replace("Final:", "").strip()
    answers.append(answer)
    print(f"Run {i+1}: {answer}")

votes = Counter(answers)
best, count = votes.most_common(1)[0]
print(f"\nVotes: {dict(votes)}")
print(f"Self-consistency answer: {best} ({count}/5 votes)")

```
