```python
from dotenv import load_dotenv
import os
from openai import OpenAI


load_dotenv()

client  = OpenAI()

model = 'gpt-4o-mini'

def ask(prompt,system=None,temperature = 0):
    """Send One prompt and return the reply text. temperature=0 keep answere stable."""
    message = []
    if system:
        message.append({"role":"system","content":system})
    message.append({"role":"user","content":prompt})
    response = client.chat.completions.create(
            model=model,messages=message,temperature=temperature
        )

    return response.choices[0].message.content.strip()



def few_shot():
    print("\n" + "=" * 60)
    print("1b. Few-Shot (for camparsion)")
    print("=" * 60)
    prompt = """Classify Each review. Answere with one word only
    Review = "Fast deilvery, great quality" -> Postive
    Review = "It broke in two days" -> Negative
    Review = "It's okay, nothing special" -> Neutral
    Review = "The delivery was late and the packaging was torn" -> 
    """

    print("Anwere:",ask(prompt))
    print("Notice:after seeing examples,the models follows the same one-word format")
    

```
