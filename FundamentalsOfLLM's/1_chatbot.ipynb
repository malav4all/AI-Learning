```python
## 1. A Model, Stripped to Its Essence

#Forget the chat window and personalities for a moment. At its core, a language model is one thing: a function that receives some text and returns some text. It runs **once**, generates its answer, and retains **nothing**.

# Ask it something a second time, and it has no idea that it has ever interacted with you before.
```

```python
def stateless_reply(prompt:str)-> str:
    return f"(Some llm response mincking the model's output) {prompt}"
```

```python
print(stateless_reply("Who is the PM of UK"))
```

```python
def stateless_chatget_demo(prompt:str) -> str:
    from dotenv import load_dotenv
    from openai import OpenAI
    load_dotenv()

    client = OpenAI()

    response = client.responses.create(
        model="gpt-4o-mini",
        input=prompt,
    )

    return response.output_text
```

```python
stateless_chatget_demo("who was the the PM of India beafor 2020")
```

```python
stateless_chatget_demo("who was before him?")
```
