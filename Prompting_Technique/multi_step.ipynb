```python
from dotenv import load_dotenv
import os
from openai import OpenAI


load_dotenv()

client  = OpenAI()

model = 'gpt-4o-mini'

def ask(prompt,system=None,temperature = 0):
    """Send One prompt and return the reply text. temperature=0 keep answere stable."""
    message = []
    if system:
        message.append({"role":"system","content":system})
    message.append({"role":"user","content":prompt})
    response = client.chat.completions.create(
            model=model,messages=message,temperature=temperature
        )

    return response.choices[0].message.content.strip()



def multi_step():
    print("\n" + "=" * 60)
    print("3. MULTI-STEP PROMPTING")
    print("=" * 60)
 
    topic = "AI for small farmers"
 
    # Step 1: brainstorm ideas
    ideas = ask(
        f"Give 3 specific angles for a blog post about '{topic}'. "
        "Numbered list only, each line under 10 words."
    )
    print("STEP 1: Ideas\n", ideas)
 
    # Step 2: pick the first idea and outline it (you could also let the user choose)
    first_idea = ideas.splitlines()[0]
    outline = ask(f"Write a blog outline for this angle, 4 bullet points only:\n{first_idea}")
    print("\nSTEP 2: Outline\n", outline)
 
    # Step 3: turn the outline into a draft
    draft = ask(f"Write a simple 150-word blog post from this outline:\n{outline}")
    print("\nSTEP 3: Draft\n", draft)
 
    # Step 4: review and improve
    final = ask(
        "Review the blog post below. First say in 2 lines what you improved, "
        "then give the final post with a catchy title.\n\n" + draft,
        system="You are a strict editor.",
    )
    print("\nSTEP 4: Review + Final\n", final)
 
    print("\nNotice: 4 separate API calls. Each step's output became the next step's input.")
    print("If one step's result is weak, you can rerun just that step.")
```

```python
multi_step()
```
